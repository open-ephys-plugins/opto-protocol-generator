```python
from enum import Enum
from typing import List

from pathlib import Path

from pydantic import BaseModel, Field
```

```python
class LightSource(BaseModel):
    """Defines a light source used for stimulation.
    """

    name: str = Field(..., title="Name of light source (e.g., LED, laser)")
    wavelength: int = Field(..., title="Wavelength of the light source in nanometers (nm)")
    output_channel: int = Field(..., title="Output channel connected to the device (e.g., 0 for first channel, 1 for second channel, etc.)")
    input_voltages: List[float] = Field(default=[], title="Input voltages used for calibration"),
    output_powers: List[float] = Field(default=[], title="Output powers corresponding to the input voltages for calibration")
    output_units: str = Field(..., title="Units of the output power (e.g., uW, mW/mm^2, etc.)")


class Device(BaseModel):
    """Defines a device that delivers light stimulation.
    """

    name: str = Field(..., title="Name of output device (e.g., probe or laser)")
    is_np_opto: bool = Field(..., title="Indicates if the device is a Neuropixels Opto probe")
    light_sources: List[LightSource] = Field(default=[], title="List of light sources used with this device")

class Configuration(BaseModel):
    """Defines the overall configuration for light stimulation.
    """

    devices: List[Device] = Field(default=[], title="List of devices used for light stimulation")
```

```python
config = Configuration(
    devices=[
        Device(
            name="Probe A",
            is_np_opto=True,
            light_sources=[
                LightSource(
                    name="Red Laser",
                    wavelength=638,
                    output_channel=0,
                    input_voltages=[0.0, 1.0, 2.0, 3.0],
                    output_powers=[0.0, 10.0, 20.0, 30.0],
                    output_units="uW"
                ),
                LightSource(
                    name="Blue Laser",
                    wavelength=450,
                    output_channel=1,
                    input_voltages=[0.0, 1.0, 2.0, 3.0],
                    output_powers=[0.0, 5.0, 15.0, 25.0],
                    output_units="uW"
                )
            ]
        ),
        Device(
            name="External Laser",
            is_np_opto=False,
            light_sources=[
                LightSource(
                    name="Green Laser",
                    wavelength=520,
                    output_channel=2,
                    input_voltages=[0.0, 1.0, 2.0, 3.0],
                    output_powers=[0.0, 8.0, 16.0, 24.0],
                    output_units="mW"
                )
            ]
        )
    ]
)
```

```python
# 1. Generate JSON string
json_data = config.model_dump_json(indent=4)

# 2. Write to file
Path("config.json").write_text(json_data)
```
